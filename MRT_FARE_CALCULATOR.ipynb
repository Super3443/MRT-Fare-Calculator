```python
def calculate_fare(distance):
    if distance <= 3.2:
        return 0.52
    elif 3.2 < distance <= 4.2:
        return 0.57
    elif 4.2 < distance <= 5.2:
        return 0.63
    elif 5.2 < distance <= 6.2:
        return 0.68
    elif 6.2 < distance <= 7.2:
        return 0.71
    else:
        return 0.74

def calculate_distance(start_station, end_station, line, dist):
    try:
        a, b = line.index(start_station), line.index(end_station)
        if a < b:
            distance = sum(dist[a:b])
            best_total_stops = b - a
        else:
            distance = sum(dist[b:a])
            best_total_stops = a - b
        
        speed = 40
        stop_time = 0.5
    
        travel_time = (distance / speed) * 60  
        stop_time = best_total_stops * stop_time
        total_time = travel_time + stop_time
        
        if a < b:
            route = line[a:b+1]
        else:
            route = list(reversed(line[b:a+1]))

        
        return distance, best_total_stops, total_time, route
    
    except ValueError:
        return None

def calculate_interchange_distance(start_station, end_station, nsl, ewl, nsl_dist, ewl_dist):

    interchanges = ["City Hall", "Raffles Place", "Jurong East"]

    if start_station in nsl:
        start_line = nsl
        start_distance = nsl_dist
    else:
        start_line = ewl
        start_distance = ewl_dist

    if end_station in nsl:
        end_line = nsl
        end_distance = nsl_dist
    else:
        end_line = ewl
        end_distance = ewl_dist

    shortest_distance = None
    best_interchange = None
    best_route = []
    
    for interchange in interchanges:
        if interchange in start_line and interchange in end_line:
            a, b = start_line.index(start_station), start_line.index(interchange)
            if a < b:
                route_1 = sum(start_distance[a:b])
                stops_1 = b - a
            else:
                route_1 = sum(start_distance[b:a])
                stops_1 = a - b
            
            c, d = end_line.index(interchange), end_line.index(end_station)
            if c < d:
                route_2 = sum(end_distance[c:d])
                stops_2 = d - c
            else:
                route_2 = sum(end_distance[d:c])
                stops_2 = c - d
            
            total_stops = stops_1 + stops_2
            total_distance = route_1 + route_2

        if shortest_distance == None or shortest_distance > total_distance:
            shortest_distance = total_distance
            best_interchange = interchange 
            best_total_stops = total_stops
            
            if a < b:
                full_route_1 = start_line[a:b+1]
            else:
                full_route_1 = list(reversed(start_line[b:a+1]))

            if c < d:
                full_route_2 = end_line[c+1:d+1]
            else:
                full_route_2 = list(reversed(end_line[d:c]))

            route = full_route_1[:-1] + [f"{start_line[b]} (Interchange)"] + full_route_2
            
        speed = 40
        stop_time = 0.5
        interchange_time = 5
    
        travel_time = ((route_1 + route_2) / speed * 60)
    
        stop_time = (best_total_stops * stop_time) + interchange_time
    
        total_time = travel_time + stop_time
    
    return shortest_distance, best_total_stops, best_interchange, total_time, route      

def station_list(line, stations, starting_list=0):
    print(f"{line} Stations:")
    print()
    for i, station in enumerate(stations, starting_list):
        print(f"{i}. {station}")
        

while True:
    try:
        nsl = ["Jurong East","Bukit Batok","Bukit Gombak","Choa Chu Kang","Yew Tee","Kranji","Marsiling","Woodlands",
                   "Admiralty","Sembawang","Canberra","Yishun","Khatib","Yio Chu Kang","Ang Mo Kio","Bishan","Braddell",
                   "Toa Payoh","Novena","Newton","Orchard","Somerset","Dhoby Ghaut","City Hall","Raffles Place","Marina Bay",
                   "Marina South Pier"]            
            
        ewl = ["Pasir Ris","Tampines","Simei","Tanah Merah","Bedok","Kembangan","Eunos","Paya Lebar","Aljunied","Kallang",
                   "Lavender","Bugis","City Hall","Raffles Place","Tanjong Pagar","Outram Park","Tiong Bahru","Redhill",
                   "Queenstown","Commonwealth","Buona Vista","Dover","Clementi","Jurong East","Chinese Garden","Lakeside",
                   "Boon Lay","Pioneer","Joo Koon","Gul Circle","Tuas Crescent","Tuas West Road","Tuas link"]

        nsl_dist = [2.1, 1.2, 3.3, 1.4, 4.1, 1.7, 1.5, 1.7, 2.4, 1.5, 1.7, 1.4, 4.9, 1.5, 2.4, 1.2,
                        0.9, 1.5, 1.2, 1.2, 1.0, 0.8, 1.0, 1.0, 1.0, 1.4]
            
        ewl_dist = [2.4, 1.4, 2.5, 1.9, 2.0, 1.1, 1.1, 1.2, 1.4, 1.1, 1.1, 1.0, 1.0, 1.2, 1.0, 1.5, 1.2, 1.4,
                        1.2, 1.1, 1.4, 1.7, 3.5, 1.5, 1.4, 1.8, 0.9, 2.6, 2.3, 1.7, 1.4, 1.3]

        interchanges = ["City Hall", "Raffles Place", "Jurong East"]

        all_stations = nsl + ewl

        discount_days = ["Friday","Saturday","Sunday"]
        non_discount_days = ["Monday", "Tuesday", "Wednesday", "Thursday"]

        print("---Welcome To The MRT Fare Calculator---")
        print()
        station_list("North-South Line", nsl, starting_list=1)
        print()
        print()
        station_list("East-West Line", ewl, starting_list=len(nsl) + 1)
        print()
            
        start = int(input("Please Choose Your Desired STARTING Point (or type 0 To Exit): "))
        if start == 0:
            print()
            print("Thank You For Using The MRT Fare Calculator!")
            print("Exiting Program")
            break
            
        else:
            start -= 1
            end = int(input("Please Choose Your Desired DESTINATION Point: "))
            end -= 1
            print()

            day = input("Enter Day Of Travel (e.g. Monday): ").strip().title()

        if not (0 <= start < len(all_stations)) or not (0 <= end < len(all_stations)):
            print("Invalid station numbers. Try again.")
            continue
            
        start_station = all_stations[start]
        end_station = all_stations[end]
            
        if start_station in nsl and end_station in nsl:
            distance, best_total_stops, total_time, route = calculate_distance(start_station, end_station, nsl, nsl_dist)
            best_interchange = None
        elif start_station in ewl and end_station in ewl:
            distance, best_total_stops, total_time, route = calculate_distance(start_station, end_station, ewl, ewl_dist)
            best_interchange = None
        else:
            distance, best_total_stops, best_interchange, total_time, route = calculate_interchange_distance(start_station, end_station, nsl, ewl, nsl_dist, ewl_dist)

        if distance is None:
            print("No Valid Route Found")
            continue

        fare = calculate_fare(distance)

        if day in discount_days:
            fare *= 0.8
            print("Discount Applied ! (20% off)")
            print()
        elif day in non_discount_days:
            print("No Discount Applied")
            print()
        else:
            print("Invalid Input. Please Try Again")
            print()
            continue
            
        if start_station in nsl:
            starting_line = "NSL"
        else:
            starting_line = "EWL"
                
        if end_station in nsl:
            ending_line = "NSL"
        else:
            ending_line = "EWL"
                
        print(f"=== TRIP SUMMARY ===")
        print(f"Trip: {start_station} → {end_station}")
        if best_interchange is not None:
            print(f"Interchange: {best_interchange} ({starting_line} → {ending_line})")
        print(f"Total Number Of Stops: {best_total_stops}")
        print(f"Estimated Time: {total_time:.0f} mins")
        print(f"Total Distance: {distance:.1f} Km")
        print(f"Total Fare: ${fare:.2f}")
            
        while True:
            print()
            show_route = input("Would You Like To View Route? (y/n): ").strip()
                
            if show_route == "y":
                print()
                print("The Entire Route Of The Journey:")
                print(" → ".join(route))
                print()
                break
            elif show_route == "n":
                print()
                break
            else:
                print()
                print("Invalid Input. Please Enter 'y' or 'n'.")

        
        exit = input("Another Trip? (y/n): ").strip()
            
        if exit == "y":
            print()
            continue
        elif exit == "n":
            print()
            print("Thank You For Using The MRT Fare Calculator!")
            print("Exiting Program")
            break
        else:
            print()
            print("Invalid Input. Please Enter 'y' or 'n'.")
        
            
    except (ValueError, IndexError):
        print()
        print("Invalid Input. Please Enter Valid Station Numbers.")
        print()
    except KeyboardInterrupt:
        print()
        print("Exiting Program.")
        break
```
